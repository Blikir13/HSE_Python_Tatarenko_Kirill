## Задание 1

```python
import re

allowed_letters="АВЕКМНОРСТУХ"

def check_car_number(car_id):
    letters_pattern = f"[{allowed_letters}]"
    pattern = rf'^({letters_pattern})(\d{{3}})({letters_pattern}{{2}})(\d{{2,3}})$'
    
    match = re.match(pattern, car_id)
    if match:
        number = ''.join(match.groups()[:3])
        region = match.group(4)
        print(f"Номер {number} валиден. Регион: {region}.")
    else:
        print("Номер не валиден.")


check_car_number('А222ВС96')
check_car_number('АБ22ВВ193')
```

```python
def remove_duplicates(text):
    pattern = r'\b(\w+)\b(?:\s+\1\b)+'
    result = re.sub(pattern, r'\1', text, flags=re.IGNORECASE)
    return result

some_string = 'Напишите функцию функцию, которая будет будет будет будет удалять все все все все последовательные повторы слов из из из из заданной строки строки при помощи регулярных выражений'
cleaned = remove_duplicates(some_string)
print(cleaned)
```
