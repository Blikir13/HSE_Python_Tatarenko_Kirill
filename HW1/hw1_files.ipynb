# Задание 1

```python
import json
import csv

dict = {}

with open('purchase_log.txt','r',encoding = 'utf-8') as f:
    f.readline()
    for line in f:
        line = line.strip()
        dictionary = json.loads(line)
        dict[dictionary['user_id']] = dictionary['category']
```

Вывод результата

```python
count = 0
for user_id, category in dict.items():
    print(f"{user_id} '{category}'")
    count += 1
    if count == 2:
        break
```

# Задание 2

```python
with open('visit_log (2).csv', 'r', encoding='utf-8') as visits_file:
    reader = csv.DictReader(visits_file)

    with open('funnel.csv', 'w', newline='', encoding='utf-8') as funnel_file:
        fieldnames = ['user_id', 'source', 'category']
        writer = csv.DictWriter(funnel_file, fieldnames=fieldnames)
        writer.writeheader()

        for row in reader:
            user_id = row['user_id']
            if user_id in dict:
                row['category'] = dict[user_id]
                writer.writerow(row)
```

Вывод результата

```python
with open('funnel.csv', encoding='utf-8') as f:
    for i, line in enumerate(f):
        print(line.strip())
        if i == 2:
            break
```
